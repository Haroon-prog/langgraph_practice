```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
# creating folder/namespace to store memory
namespace = ('user','u1')
namespace2 = ('user','u2')

# user
# |_u1
# |
# |_u2
```

```python
# creating memory

store.put(namespace,"1",{"data":"user likes pizza"})
store.put(namespace,'2',{'data':'user prefer python'})
```

```python
# retriving particular memory from the memory

store.get(namespace,"1")
```

```python
# to retrive all memories from particular folder (eg u1)

items = store.search(namespace) # return list of dict

for item in items:
    print(item)
```
