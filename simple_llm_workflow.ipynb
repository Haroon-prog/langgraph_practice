```python
from langgraph.graph import StateGraph,START,END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv

load_dotenv()
```

```python
# creating state for workflow 
class llmState(TypedDict):
    question : str
    answer : str 
```

```python
model = ChatGoogleGenerativeAI(model='gemini-2.5-flash')


# creating a function 
def llm_model(state : llmState) -> llmState :
    #extract question
    question = state['question']

    # creating prompt 
    prompt = f'answer the following question briefly {question}'

    #giving prompt to model 
    answer = model.invoke(prompt).content

    #update answer in state 
    state['answer'] = answer

    return state
```

```python
graph = StateGraph(llmState)

# adding node 
graph.add_node('llm_model', llm_model)

# adding edges 
graph.add_edge(START,'llm_model')
graph.add_edge('llm_model',END)

# compile
workflow = graph.compile()
```

```python
initial_state = {'question': 'what is the distance from US to india ?' }
final_state = workflow.invoke(initial_state)
```
